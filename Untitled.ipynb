```python
import openstudio
import pandas as pd
from openstudio.openstudiomodel import *


# Get all IddFields for an IddObject
#
# @param iddObject [OpenStudio::IddObject] the IddObject to scan
# @return [Array[OpenStudio::IddField]
def get_idd_fields(iddObject: "openstudio.IddObject") -> List["openstudio.IddField"]:
    """Get all IddFields for an IddObject.

    Parameters
    ----------
    iddObject : openstudio.IddObject
        the IddObject to scan

    Returns
    -------
    fields : List[openstudio.IddField]
    """
    num_fields = iddObject.numFields() + iddObject.properties().numExtensible
    return [iddObject.getField(i).get() for i in range(num_fields)]

def get_fields(iddObject: "openstudio.IddObject") -> Dict[str, int]:
    """Scans all fields for an IddObject and return a dict of name to index.

    Parameters
    ----------
    iddObject : openstudio.IddObject
        the IddObject to scan

    Returns
    -------
    fields : Dict[str, int]
        keys are the field names, values are the indices (0-indexed)
    """
    result = {}
    for i, f in enumerate(get_idd_fields(iddObject=iddObject)):
        result[f.name()] = i
    return result



factory = openstudio.IddFileAndFactoryWrapper(openstudio.IddFileType("OpenStudio"))

os_building = factory.getObject("OS:Building").get()
print(os_building)

os_space = factory.getObject("OS:Space").get()
print(os_space)
```

```python
import geomeffibem
floor_sf = geomeffibem.Surface.Floor()
floor_sf.to_Point3dVector()
```

```python
m = openstudio.model.Model()
```

```python
c = openstudio.model.Construction(m)
c.setName("R13 Construction")
mat = openstudio.model.MasslessOpaqueMaterial(m)
mat.setName("R13-IP")
mat.setThermalResistance(openstudio.convert(13, "ft^2*h*R/Btu","m^2*K/W").get())
c.setLayers([mat])
```

```python
simple_glazing = openstudio.model.SimpleGlazing(m)
simple_glazing.setName("Simple Glazing Mat")
window_cons = openstudio.model.Construction(m)
window_cons.setName("Simple Glazing")
window_cons.setLayers([simple_glazing])
```

```python
z = openstudio.model.ThermalZone(m)
z.setName("Zone1")
space = Space.fromFloorPrint(floor_sf.to_Point3dVector(), 3.0, m, "Space1").get()
space.setThermalZone(z)
```

```python
[s.setConstruction(c) for s in m.getSurfaces()]
```

```python
roof = next(s for s in space.surfaces() if s.surfaceType().lower() == 'roofceiling')

vertices = roof.vertices()
g = openstudio.getCentroid(vertices).get()
scale_factor = 0.4**0.5  # sqrt(40%)

new_vertices = []
for vertex in vertices:
    # Point3d - Point3d = Vector3d
    # Vector from centroid to vertex (GA, GB, GC, etc)
    centroid_vector = vertex - g
    
    # Resize the vector (done in place) according to scale_factor
    centroid_vector.setLength(centroid_vector.length() * scale_factor)
    
    # Move the vertex toward the centroid
    vertex = g + centroid_vector
    
    new_vertices.append(vertex)
ss = openstudio.model.SubSurface(new_vertices, m)
ss.setName("Skylight")
ss.setConstruction(window_cons)
ss.setSurface(roof)
```

```python
walls = sorted([s for s in space.surfaces() if s.surfaceType().lower() == 'wall'], key=lambda s: s.nameString())
assert len(walls) == 4
```

```python
for w, ss_type in zip(walls[:-1], ["FixedWindow", "OperableWindow", "Window"]):
    ss = w.setWindowToWallRatio(0.4).get()
    ss.setSubSurfaceType(ss_type)
    ss.setName(f"{ss_type} SubSurface")
    ss.setConstruction(window_cons)
```

```python
w = walls[-1]
print(w)
```

```python
import matplotlib.pyplot as plt
```

```python
wall = geomeffibem.Surface.from_Surface(w)
assert wall.get_plot_axis() == 'xz'
xsfs = wall.split_into_n_segments(n_segments=3, axis='x')

new_doors = []

v_np = xsfs[0].to_numpy()
min_x = v_np[:, 0].min()
max_x = v_np[:, 0].max()

min_z = v_np[:, 2].min()
max_z = v_np[:, 2].max()

y = v_np[0, 1]

overhead_door = geomeffibem.Surface.Rectangle(min_x=min_x + 0.5, max_x=max_x - 0.5, min_y=y, max_y=y, min_z=min_z, max_z=max_z - 0.5)
overhead_door.name = "OverHead Door"
new_doors.append(overhead_door)

v_np = xsfs[1].to_numpy()
min_x = v_np[:, 0].min()
max_x = v_np[:, 0].max()
center_x = min_x + (max_x - min_x) / 2

min_z = v_np[:, 2].min()

y = v_np[0, 1]

door = geomeffibem.Surface.Rectangle(min_x=center_x - 0.4, max_x=center_x + 0.4, min_y=y, max_y=y, min_z=min_z, max_z=min_z + 2.0)
door.name = "Door"
new_doors.append(door)

v_np = xsfs[2].to_numpy()
min_x = v_np[:, 0].min()
max_x = v_np[:, 0].max()
center_x = min_x + (max_x - min_x) / 2

min_z = v_np[:, 2].min()

y = v_np[0, 1]

door = geomeffibem.Surface.Rectangle(min_x=center_x - 0.4, max_x=center_x + 0.4, min_y=y, max_y=y, min_z=min_z, max_z=min_z + 2.0)
door.name = "GlassDoor"
new_doors.append(door)

ax = wall.plot()

for new_sf in xsfs:
    new_sf.plot(ax=ax, annotate=False) #, with_os_centroid=True)

for new_sf in new_doors:
    new_sf.plot(ax=ax, annotate=False) #, with_os_centroid=True)
```
